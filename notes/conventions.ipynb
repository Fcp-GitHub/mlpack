# Conventions Used in These Notes
In these notes I've tried to be as consistent as possible with my notation. 

Here I try to list all the conventions used, which have to be assumed as true in all the notes files unless otherwise specified.

# Disclaimer
I try to use the following matrix/vector notation (e.g. samples' matrix instead of single matrix vectors) for two main reasons:
1. It is a compact way of writing the entire calculations on all datasets.
2. They are the actual formulas that are used in the package's implementation as well.

# Fundamental Concepts
- Labels' vector: $\boldsymbol y$ .
- Predicted labels' vector during training: $\boldsymbol y'$.
- Samples' matrix: $X$ .
- Number of samples: $S$.
- Weights' vector: $\boldsymbol w$ .
- Learning rate: $\eta$ .
- Loss function: $l(\dots)$ .

# Mathematical Concepts
- Heaviside step function: $H(x)$.
  <div class = "alert alert-block alert-info">
      The Heaviside step function is here defined as a <b>piecewise function</b>:
      $$
      H(x) \coloneqq \begin{cases} 1, & x \geq 0 \\ 0, & x \lt 0 \end{cases}
      $$
