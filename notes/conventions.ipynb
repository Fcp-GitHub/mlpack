# Conventions Used in These Notes
In these notes I've tried to be as consistent as possible with my notation. 

Here I try to list all the conventions used, which have to be assumed as true in all the notes files unless otherwise specified.

# Disclaimer
I try to use the following matrix/vector notation (e.g. samples' matrix instead of single matrix vectors) for two main reasons:
1. It is a compact way of writing the entire calculations on all datasets.
2. They are the actual formulas that are used in the package's implementation as well.

# Fundamental Concepts
- Labels' vector: $\boldsymbol y$ (for binary classification).
- Labels' matrix of size $(\mathrm{num\_samples}\times \mathrm{num\_classes})$: $Y$ (for multiclass classification).
- Predicted labels' vector during training: $\boldsymbol y'$.
- Predicted labels' matrix of size $(\mathrm{num\_samples}\times \mathrm{num\_classes})$: $Y'$ (for multiclass classification).
- Samples' matrix: $X$ .
- Number of samples: $S$.
- Weights' vector: $\boldsymbol w$ (for binary classification).
- Weights' matrix of size $(\mathrm{num\_samples}\times \mathrm{num\_classes})$: $W$ (for multiclass classification).
- Learning rate: $\eta$ .
- Loss function: $l(\dots)$ .

# Abbreviations
- In any formula where a linear combination of samples and weights appears there should also be a bias term summed. However, for simplicity it is not added in the notes.

# Mathematical Concepts
- Heaviside step function: $H(x)$.
